# Reduce metric variance with CUPED

[All How-to Guides](README.md)

You have a post-treatment outcome and a related historical metric for the same users. CUPED uses the historical values to reduce outcome variance. Check that the historical column was measured before treatment.

## Set up the comparison

1. Assign `TargetRole` to the outcome and `PreTargetRole` to the historical metric.
2. Map the outcome to its historical covariate in `cuped_features`.
3. Compare the original and adjusted rows and read the separate variance reduction report.

`PreTargetRole()` describes the historical column. The mapping in `cuped_features` is sufficient for CUPED. CUPAC uses `parent` and `lag` to describe the historical periods.

```python
from hypex import ABTest
from hypex.dataset import Dataset, PreTargetRole, TargetRole, TreatmentRole
from hypex.utils.tutorial_data_creation import gen_control_variates_df

frame = gen_control_variates_df(
    data_size=1200, dependent_division=False, random_state=42
)
data = Dataset(
    roles={
        "Treatment": TreatmentRole(),
        "Target": TargetRole(),
        "Target_lag_1": PreTargetRole(),
    },
    data=frame,
)
result = ABTest(cuped_features={"Target": "Target_lag_1"}).execute(data)
result.summary
```

## Read the variance reduction report

```python
result.variance_reduction_report
```

## Interpret the result

In `summary`, find `Target` and `Target_cuped`. In `variance_reduction_report`, `variance_reduction_pct` shows a reduction of about 10%. To draw a conclusion about the effect, use the estimate and test for the outcome specified in your analysis plan.
