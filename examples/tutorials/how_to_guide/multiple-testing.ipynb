# Adjust for multiple comparisons in an A/B test

[All How-to Guides](README.md)

Suppose a significant result on either of two metrics would affect a decision. Define the family of hypotheses in advance and use the Holm adjustment. It controls the probability of at least one false rejection in the family.

## Set up the comparison

1. Assign `TreatmentRole` to the group column and `TargetRole` to both post-treatment metrics. In this example, `Target` is the primary outcome and `X` is the secondary outcome.
2. Run a single A/B test with both metrics and open the adjustment report.

The two columns with `TargetRole` form the family. `multitest_method="holm"` explicitly selects the default adjustment.

```python
from hypex import ABTest
from hypex.dataset import Dataset, TargetRole, TreatmentRole
from hypex.utils.tutorial_data_creation import gen_control_variates_df

frame = gen_control_variates_df(
    data_size=1200, dependent_division=False, random_state=42
)
data = Dataset(
    roles={"Treatment": TreatmentRole(), "Target": TargetRole(), "X": TargetRole()},
    data=frame,
)
result = ABTest(multitest_method="holm").execute(data)
result.summary
```

## Open the adjustment report

```python
result.multitest
```

## Interpret the results

Use the difference in means from `summary` as the effect estimate. In `multitest`, `new p-value` gives the adjusted p-value and `rejected` gives the decision. For `Target`, the p-value moves from about 0.29 to 0.58. Neither hypothesis is rejected at 0.05. If there is just one group contrast and one metric, `multitest` returns a text message.
